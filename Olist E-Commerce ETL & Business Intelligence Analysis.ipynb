#### Inserting New Columns ( Feature Engineering)

```python
ocd_orders['delivery_days'] = (
    ocd_orders['order_delivered_customer_date']
    - ocd_orders['order_purchase_timestamp']
).dt.days
```

```python
ocd_orders.head(10)
```

```python
ocd_orders['delivery_delay'] = (
    ocd_orders['order_delivered_customer_date']
    - ocd_orders['order_estimated_delivery_date']
).dt.days 
```

```python
ocd_orders.head(10)
```

```python
ocd_orders['delivery_status'] = np.where(
    ocd_orders['delivery_delay'] > 0,
    'Late',
    'On Time'
)
```

```python
ocd_orders['delivery_status'].head(10)
```

```python
ocd_order_items.head()
```

```python
ocd_order_items['order_value'] = (
    ocd_order_items['price'] +
    ocd_order_items['freight_value']
)
```

```python
ocd_order_items.head()
```

```python
order_value=ocd_order_items.groupby('order_id', as_index=False)\
    .agg(order_value=('order_value', 'sum'))
```

```python
ocd_orders =ocd_orders.merge(
    order_value,
    on='order_id',
    how='left'
)
```

#### Exportin files to sql

```python
from sqlalchemy import create_engine
```

```python
engine = create_engine(
    "postgresql+psycopg2://postgres:Dhiman123@localhost:5432/olist_db"
)
```

```python
with engine.connect() as conn:
    print("PostgreSQL connection successful!")
```

```python
tables = {
    "ocd_orders": ocd_orders,
    "ocd_order_items": ocd_order_items,
    "ocd_customers": ocd_customers,
    "ocd_products": ocd_products,
    "ocd_sellers": ocd_sellers,
    "ocd_order_payments": ocd_order_payments,
    "ocd_order_reviews": ocd_order_reviews,
    "ocd_geolocation": ocd_geolocation,
    "ocd_product_category" :ocd_product_category
}
```

```python
for table_name, df in tables.items():
    df.to_sql(
        table_name,
        engine,
        if_exists="replace",
        index=False
    )
    print(f"{table_name} loaded successfully: {len(df)} rows")
```
